```python
class Kiosk:
    def __init__(self):
       
        self.menu = ['americano', 'latte', 'mocha', 'yuza_tea', 'green_tea', 'choco_latte']
        self.price = [2000, 3000, 3000, 2500, 2500, 3000]
        
        self.order_menu = []
        self.order_price = []

    # 메뉴 출력 메서드
    def menu_print(self):
        print("\n" + "="*30)
        print("      AIFFEL CAFE MENU")
        print("="*30)
        for i in range(len(self.menu)):
            print(f"  {i+1}. {self.menu[i].ljust(12)} : {self.price[i]}원")
        print("="*30)

    # 주문 메서드
    def menu_select(self):
        while True: 
            n = 0
            while n < 1 or n > len(self.menu):
                try:
                    n = int(input("\n음료 번호를 입력하세요 : "))
                except ValueError:
                    print("숫자만 입력 가능합니다.")
                    continue

                if 1 <= n <= len(self.menu):
                   
                    t = 0
                    while t not in [1, 2]:
                        try:
                            t = int(input("HOT(1) 또는 ICE(2)를 선택하세요 : "))
                        except ValueError:
                            print("1 또는 2를 입력해주세요.")
                            continue
                        
                        if t == 1: self.temp = "HOT"
                        else: self.temp = "ICE"
                    
                   
                    selected_item = f"{self.temp} {self.menu[n-1]}"
                    selected_price = self.price[n-1]
                    
                    self.order_menu.append(selected_item)
                    self.order_price.append(selected_price)
                    
                    print(f"-> 장바구니에 담겼습니다: {selected_item} ({selected_price}원)")
                else:
                    print("없는 메뉴입니다. 다시 주문해 주세요.")

           
            additional = input("\n추가 주문하시겠습니까? (y 입력 시 추가 주문 / n 입력 시 결제) : ").lower()
            if additional == 'n':
                print("주문을 종료하고 결제로 이동합니다.")
                break  
            elif additional == 'y':
                print("추가 주문을 진행합니다.")
            else:
                print("잘못된 입력입니다. 결제 단계로 넘어갑니다.")
                break

    # 지불 메서드
    def pay(self):
        self.total_sum = sum(self.order_price)
        print(f"\n총 결제 금액은 {self.total_sum}원입니다.")
        
        while True:
            payment_method = input("결제방법 선택 (1.현금 / 2.카드) : ")
            if payment_method in ['1', '현금']:
                print('직원을 호출하겠습니다.')
                break
            elif payment_method in ['2', '카드']:
                print('IC칩 방향에 맞게 카드를 꽂아주세요.')
                break
            else:
                print('다시 결제방법을 선택해 주세요.')

    # 주문서(영수증) 출력 
    def table(self):
        print('\n' + '⟝' + '-' * 30 + '⟞')
        print(f"| {'[ 영 수 증 ]'.center(26)} |")
        print('|' + '-' * 30 + '|')
        
        for i in range(len(self.order_menu)):
            line = f"{self.order_menu[i]} : {self.order_price[i]}원"
            print(f"| {line.ljust(28)} |")
            
        print('|' + '-' * 30 + '|')
        print(f"| {'총 합계 : ' + str(sum(self.order_price)) + '원':<28} |")
        print('⟝' + '-' * 30 + '⟞')


a = Kiosk()
a.menu_print()
a.menu_select()
a.pay()
a.table()
```

```python
class Square:
    def __init__(self):
        self.square = int(input('넓이를 구하고 싶은 사각형의 숫자를 써주세요.\n 1.직사각형 2.평행사변형 3.사다리꼴 \n >>>'))

        if self.square == 1:
            print('직사각형 함수는 rect()입니다.')
        elif self.square == 2:
            print('평행사변형 함수는 par()입니다.')
        elif self.square == 3:
            print('사다리꼴 함수는 trape()입니다.')
        else:
            print('1, 2, 3 중에서 다시 입력해주세요')

    def rect(self):
        width, vertical = map(int, input('가로, 세로를 입력하세요. 예시 : 가로,세로\n >>>').split(','))
        area = width * vertical
        result = '직사각형의 넓이는 : ' + str(area)
        return result

    def par(self):
        base, height = map(int, input('밑변, 높이를 입력하세요. 예시 : 밑변,높이\n >>>').split(','))
        area = base * height
        result = '평행사변형의 넓이는 : ' + str(area)
        return result

    def trape(self):
        
        top, bottom, height = map(int, input('윗변, 아랫변, 높이를 입력하세요. 예시 : 윗변,아랫변,높이\n >>>').split(','))
        area = (top + bottom) * height / 2
        result = '사다리꼴의 넓이는 : ' + str(area)
        return result


a = Square()
```

```python
a.trape()
```
